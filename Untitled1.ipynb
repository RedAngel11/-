```python
import numpy as np
import matplotlib.pyplot as plt

raw = {
    3:  [4.92, 4.93, 5.07],
    4:  [6.68, 6.48, 6.32],
    5:  [8.25, 8.35, 8.30],
    6:  [9.55, 9.12, 9.49],
    7:  [11.55, 11.20, 11.51],
    8:  [12.98, 12.69, 12.57],
    9:  [15.84, 16.07, 16.01],
    10: [16.56, 17.00, 16.59],
    11: [19.36, 20.12],        # 15.36 отброшено как описка
    12: [20.94, 20.00, 20.32],
}
d_instr = 0.0   # инструментальная погрешность секундомера, с;
                # поставь напр. 0.1, если преподаватель требует её учесть

n_arr, T_arr, dT_arr = [], [], []
for n, ts in raw.items():
    ts = np.array(ts)
    s = ts.std(ddof=1)
    n_arr.append(n)
    T_arr.append(ts.mean()/5)
    dT_arr.append(np.sqrt((s/np.sqrt(len(ts)))**2 + d_instr**2)/5)
n_arr, T_arr, dT_arr = map(np.array, (n_arr, T_arr, dT_arr))

for n, T, dT in zip(n_arr, T_arr, dT_arr):
    print(f"n = {n:>2}:  T = {T:.2f} ± {dT:.2f} с")

# МНК со свободным членом + вклад времени в δk
N, n_bar, T_bar = len(n_arr), n_arr.mean(), T_arr.mean()
Snn = ((n_arr-n_bar)**2).sum()
k = ((n_arr-n_bar)*(T_arr-T_bar)).sum()/Snn
a = T_bar - k*n_bar
res = T_arr - (a + k*n_arr)
sig = np.sqrt((res**2).sum()/(N-2))
dk, da = sig/np.sqrt(Snn), sig*np.sqrt(1/N + n_bar**2/Snn)
dk_T = 1/np.sqrt((((n_arr-n_bar)**2)/dT_arr**2).sum())
print(f"k = {k:.4f} ± {dk:.4f} с (МНК), вклад времени δk_T = {dk_T:.4f} с, "
      f"итого {np.hypot(dk, dk_T):.4f} с;  a = {a:.3f} ± {da:.3f} с")

# График
plt.rcParams.update({'figure.dpi': 110, 'font.size': 11, 'axes.linewidth': 0.8,
                     'xtick.direction': 'in', 'ytick.direction': 'in',
                     'xtick.major.size': 4, 'ytick.major.size': 4,
                     'xtick.minor.size': 2.5, 'ytick.minor.size': 2.5})
fig, ax = plt.subplots(figsize=(8, 5.2))
nn = np.linspace(2.5, 12.5, 200)

ax.errorbar(n_arr, T_arr, yerr=dT_arr, fmt='o', ms=5, color='#1399EB',
            ecolor='#1399EB', elinewidth=1.2, capsize=3, label='значение')
ax.plot(nn, a + k*nn, '-', color='#8B20D4', lw=2,
        label=f'МНК: $k = {k:.4f} \\pm {dk:.4f}$ с, $a = {a:.2f} \\pm {da:.2f}$ с')

ax.set_xlim(2.4, 12.6)
ax.set_ylim(0.6, 4.4)
ax.minorticks_on()
ax.grid(True, which='major', alpha=0.5, lw=0.7)
ax.grid(True, which='minor', alpha=0.2, lw=0.4)
ax.set_xlabel(r'$n$, число шариков')
ax.set_ylabel(r'период $T = t_5/5$, с')
ax.set_title('Крутильные колебания стрелки: зависимость $T(n)$')
ax.legend(loc='upper left', fontsize=10, framealpha=0.9)

plt.tight_layout()
plt.savefig('T_vs_n.pdf')   # векторный — прямо в Beamer
plt.savefig('T_vs_n.png', dpi=200)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ---------- Данные задания 3 ----------
n    = np.array([4, 6, 8, 10, 12], float)
m_gr = np.array([0.803, 0.376, 0.413, 0.382, 0.429])   # масса грузика, г
k_d  = np.array([1, 2, 3, 4, 4], float)                # плечо r_gr в диаметрах d
g, d = 981.0, 0.5804                                   # см/с^2, см
dm_bal, dd = 0.05, 0.04                                # весы, г; диаметр, см

# Механический момент и его погрешность:
# дельта r_gr / r_gr = дельта d / d (плечо = k*d), плюс весы
M  = m_gr * g * k_d * d
dM = M * np.sqrt((dm_bal / m_gr)**2 + (dd / d)**2)

# ---------- МНК со свободным членом: M = b + A*n ----------
N, n_bar, M_bar = len(n), n.mean(), M.mean()
Snn = ((n - n_bar)**2).sum()
A = ((n - n_bar)*(M - M_bar)).sum() / Snn
b = M_bar - A*n_bar
res = M - (b + A*n)
sig = np.sqrt((res**2).sum() / (N - 2))
dA = sig / np.sqrt(Snn)
db = sig * np.sqrt(1/N + n_bar**2/Snn)

print('МНК: M = b + A*n')
print(f'A = {A:.1f} ± {dA:.1f} дин·см')
print(f'b = {b:.1f} ± {db:.1f} дин·см  (согласуется с нулём => аддитивность OK)')

for Pm, rel in ((67.1, 0.17), (77.5, 0.14)):
    B = A / Pm
    dB = B * np.sqrt((dA/A)**2 + rel**2)
    print(f'Pm = {Pm:5.1f}:  B_perp = {B:.2f} ± {dB:.2f} Гс')

# ---------- График в стиле T(n) ----------
plt.rcParams.update({'figure.dpi': 110, 'font.size': 11, 'axes.linewidth': 0.8,
                     'xtick.direction': 'in', 'ytick.direction': 'in',
                     'xtick.major.size': 4, 'ytick.major.size': 4,
                     'xtick.minor.size': 2.5, 'ytick.minor.size': 2.5})
fig, ax = plt.subplots(figsize=(8, 5.2))
nn = np.linspace(3, 13, 100)
ax.errorbar(n, M, yerr=dM, fmt='o', ms=5, color='#1DEB6E',
            ecolor='#1DEB6E', elinewidth=1.2, capsize=3, label='значения')
ax.plot(nn, b + A*nn, '-', color='#9E5EEB', lw=2,
        label=f'МНК: $A = {A:.1f} \\pm {dA:.1f}$ дин·см, $b = {b:.0f} \\pm {db:.0f}$ дин·см')
ax.set_xlim(3, 13)
ax.minorticks_on()
ax.grid(True, which='major', alpha=0.5, lw=0.7)
ax.grid(True, which='minor', alpha=0.2, lw=0.4)
ax.set_xlabel(r'$n$, число шариков')
ax.set_ylabel(r'механический момент $M$, дин·см')
ax.set_title('Уравновешивание стрелки: зависимость $M(n)$')
ax.legend(loc='upper left', fontsize=10, framealpha=0.9)
plt.tight_layout()
plt.savefig('M_vs_n.pdf')
plt.savefig('M_vs_n.png', dpi=200)
plt.show()
```
